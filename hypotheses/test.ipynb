📌 H2 : Доход ↔ Потребление

Делаем корреляцию (Pearson / Spearman).

```python
from scipy.stats import pearsonr, spearmanr

pearson_corr, pearson_p = pearsonr(df["revenu_moyen"], df["conso_par_habitant"])
spearman_corr, spearman_p = spearmanr(df["revenu_moyen"], df["conso_par_habitant"])

print(f"Pearson: r={pearson_corr:.2f}, p={pearson_p:.4f}")
print(f"Spearman: r={spearman_corr:.2f}, p={spearman_p:.4f}")
```

📌 H3 : Возраст жилья ↔ Потребление

Строим линейную регрессию:

```python
import statsmodels.api as sm

X = df[["age_logement"]]  # возраст жилья
y = df["conso_par_habitant"]

X = sm.add_constant(X)  # добавляем intercept
model = sm.OLS(y, X).fit()
print(model.summary())
```
