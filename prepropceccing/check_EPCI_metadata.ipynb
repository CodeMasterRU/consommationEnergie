```python
import pandas as pd
```

```python
epci_df = pd.read_csv("EPCI_metadata.csv")
```

```python
print("STRUCTURE DU FICHIER :")
print(epci_df.info())
```

```python
print("\nAPERCU DES DONNEES :")
print(epci_df.head())
```

```python
duplicate_count = epci_df.duplicated().sum()
print(f"\nDuplicates: {duplicate_count}")
```

```python
print("\n Missing values per column:")
print(epci_df.isna().sum())
```

```python
print("\n Nombre de CODE_EPCI uniques :", epci_df['CODE_EPCI'].nunique())
print("Nombre de NOM_EPCI uniques  :", epci_df['NOM_EPCI'].nunique())
```

```python
duplicates_by_code = epci_df[epci_df.duplicated(subset='CODE_EPCI', keep=False)]
if not duplicates_by_code.empty:
    print("\n Duplicates sur CODE_EPCI :")
    print(duplicates_by_code)
else:
    print("\n Aucun doublon sur CODE_EPCI")
```
